### LogP = Lipophilicity (fat-loving property)

```txt
LogP < 0
→ Water-loving (Hydrophilic)

LogP ≈ 1–3
→ Balanced

LogP > 5
→ Very fat-loving (Lipophilic)
```

````python
from rdkit import Chem
from rdkit.Chem import Descriptors

mol = Chem.MolFromSmiles("CCO")

print(Descriptors.MolLogP(mol))
````
