# Aromaticity

## Definition

**Aromaticity** is a property of certain cyclic molecules in which π-electrons are delocalized over the entire ring, making the molecule unusually stable.

## Conditions for Aromaticity (Hückel's Rule)

A molecule is aromatic if it satisfies all of the following conditions:

1. It is **cyclic** (forms a ring).
2. It is **planar** (flat).
3. It has **continuous conjugation** (overlapping p-orbitals around the ring).
4. It contains **(4n + 2) π-electrons**, where *n* = 0, 1, 2, ...

## Example: Benzene

* Molecular Formula: C₆H₆
* Number of π-electrons: 6
* Hückel's Rule: 6 = 4(1) + 2

Therefore, **benzene is aromatic**.

## Why is Benzene Aromatic?

* Benzene consists of a six-membered carbon ring.
* Each carbon atom is **sp² hybridized**.
* The six π-electrons are delocalized over the entire ring.
* This electron delocalization makes benzene highly stable.

## Benzene SMILES String

**Aromatic representation:**

```text
c1ccccc1
```

**Kekulé representation:**

```text
C1=CC=CC=C1
```

In cheminformatics tools such as RDKit and DeepChem, the aromatic form (`c1ccccc1`) is commonly used.

## Examples

| Molecule    | SMILES    | Aromatic |
| ----------- | --------- | -------- |
| Benzene     | c1ccccc1  | Yes      |
| Toluene     | Cc1ccccc1 | Yes      |
| Cyclohexane | C1CCCCC1  | No       |

## Key Point

**Aromatic molecules are more stable because their π-electrons are delocalized across the entire ring.**

### Quick Memory Tip

> **Aromaticity = Cyclic + Planar + Conjugated + (4n + 2) π-electrons**.


````python
from rdkit import Chem
from torchdrug import data
import matplotlib

%matplotlib inline

````

````python
smiles='c1ccccc1'
mol=data.Molecule.from_smiles(smiles)
mol.visualize()
````

````python
mol=Chem.MolFromSmiles(smiles)
print(mol.GetAtomWithIdx(3).GetIsAromatic())
print(mol.GetAtomWithIdx(1).GetIsAromatic())
print(mol.GetBondBetweenAtoms(3,4).GetIsAromatic())
````

````python
for bond in mol.GetBonds():
    print(bond.GetBondType())
    print(bond.GetIdx())
````
