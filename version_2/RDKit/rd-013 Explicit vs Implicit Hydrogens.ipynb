| Feature                     | Implicit H | Explicit H    |
| --------------------------- | ---------- | ------------- |
| Hydrogen visible in SMILES? |  No       | Yes         |
| Automatically calculated?   | Yes      | No          |
| Shorter notation?           | Yes      | No          |
| Common in datasets?         | Yes      | Less common |


###  Example: Ethanol
- Implicit:
```txt
CCO
```
- Actual structure
```txt
CH3-CH2-OH
```
- Explicit
```txt
[H]C([H])([H])C([H])([H])O[H]
``` 

````python
from rdkit import Chem
from torchdrug import data
import matplotlib
import cirpy
%matplotlib inline
````

````python
data.Molecule.from_smiles("CCO").visualize()
print(cirpy.resolve("ethanol","smiles"))
print(cirpy.resolve("CCO","names"))
````
