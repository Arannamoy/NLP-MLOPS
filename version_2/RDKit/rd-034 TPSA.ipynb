### TPSA (Topological Polar Surface Area)
- TPSA measures the total polar surface area of a molecule.

Low TPSA
→ Easier membrane penetration

High TPSA
→ More difficult membrane penetration


TPSA < 60
→ Good membrane permeability

TPSA < 140
→ Usually good oral absorption

TPSA > 140
→ Poor absorption likely

```python
from rdkit import Chem
from rdkit.Chem import Descriptors

mol = Chem.MolFromSmiles("CCO")

print(Descriptors.TPSA(mol))
```
