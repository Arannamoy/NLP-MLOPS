### Organic Subset Atoms (RDKit / SMILES)

**Definition:**
Organic subset atoms are common elements in SMILES that can be written **without square brackets `[]`**.

**Organic Subset Atoms:**

```text
B, C, N, O, P, S, F, Cl, Br, I
```

**Aromatic Forms:**

```text
b, c, n, o, p, s
```

### Examples

```text
CCO
```

Ethanol

```text
C-C-O
```

---

```text
c1ccccc1
```

Benzene (aromatic)

---

### No Brackets Needed

```text
CCO
CCN
CCCl
```

Because C, N, O, and Cl are organic subset atoms.

---

### Brackets Required

```text
[Na+]
[Fe]
[Zn]
[Cu+2]
[Mg]
```

Because these atoms are **not** in the organic subset.

---

### Why?

RDKit automatically knows the common valence of:

```text
C → 4
N → 3
O → 2
```

So brackets are not needed.

For metals and uncommon atoms, brackets are required.

---

### Quick Rule

```text
Organic Subset Atoms
=
B C N O P S F Cl Br I
```

- No brackets needed

```text
CCO
CCN
CCBr
```

- Brackets required

```text
[Na+]
[Fe]
[Zn]
```

### RDKit Example

```python
from rdkit import Chem

mol = Chem.MolFromSmiles("CCO")

for atom in mol.GetAtoms():
    print(atom.GetSymbol())
```

Output:

```text
C
C
O
```

**Remember:** Most drug molecules mainly contain **C, N, O, S, P, F, Cl, Br, and I**, so Organic Subset Atoms appear very frequently in cheminformatics and drug discovery.
