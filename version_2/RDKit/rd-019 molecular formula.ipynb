```python
from rdkit import Chem
from rdkit.Chem import Draw,rdMolDescriptors
import cirpy
```

```python
benzene_smiles=cirpy.resolve("benzene","smiles")
nitric_acid_smiles=cirpy.resolve("nitric acid","smiles")
nitric_oxide_smiles=cirpy.resolve("azinous acid","smiles")
```

```python
print(benzene_smiles,nitric_acid_smiles,nitric_oxide_smiles)
```

```python
mol1=Chem.MolFromSmiles(benzene_smiles)
mol2=Chem.MolFromSmiles(nitric_acid_smiles)
mol3=Chem.MolFromSmiles(nitric_oxide_smiles)
```

```python
Draw.MolsToImage([mol1,mol2,mol3])
```

```python
rdMolDescriptors.CalcMolFormula(mol1)
```

```python
rdMolDescriptors.CalcMolFormula(mol2)
```

```python
rdMolDescriptors.CalcMolFormula(mol3)
```
