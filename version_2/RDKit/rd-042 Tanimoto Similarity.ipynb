```python
from rdkit import Chem
from rdkit.Chem import Draw,rdFingerprintGenerator,RDKFingerprint,MACCSkeys
```

```python
mol=Chem.MolFromSmiles("C[C@H](N)C(=O)NCC1=CC=C(C=C1)S(=O)(=O)NCCOP(=O)([O-])OCC[NH3+]")
Draw.MolToImage(mol)
```

```python
FGen=rdFingerprintGenerator.GetMorganGenerator()
fp2=RDKFingerprint(mol)
```

```python
fp1=FGen.GetFingerprint(mol)
```

```python
from rdkit.DataStructs import TanimotoSimilarity
similarity=TanimotoSimilarity(fp1,fp2)
```

```python
print(similarity)
```
