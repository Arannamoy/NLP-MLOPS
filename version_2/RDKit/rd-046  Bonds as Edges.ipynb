```python
from rdkit import Chem
from rdkit.Chem import Draw
```

```python
mol=Chem.MolFromSmiles('C[C@H](N)C(=O)NCC1=CC=C(C=C1)S(=O)(=O)NCCOP(=O)([O-])OCC[NH3+]')
```

```python
for bond in mol.GetBonds():
    print("Begin atom: ",bond.GetBeginAtom().GetSymbol())
    print("End atom: ",bond.GetEndAtom().GetSymbol())
    print("==================================")
```
