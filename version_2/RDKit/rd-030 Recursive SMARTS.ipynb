```python
from rdkit import Chem
from rdkit.Chem import rdMolDescriptors,Draw
import cirpy
```

```python
results=cirpy.resolve("acetic acid","smiles")
print(results)
```

```python
mol=Chem.MolFromSmiles(results)
smarts=Chem.MolToSmarts(mol)
print(smarts)
```
