### Isotopes in SMILES / RDKit

**Definition:**
An isotope is an atom with the same element but a different number of neutrons.

In SMILES, isotopes are written inside **square brackets `[]`**.

---

### Examples

#### Carbon-13

```text
[13C]
```

Meaning:

```text
Carbon isotope with mass number 13
```

---

#### Deuterium (Hydrogen-2)

```text
[2H]
```

Meaning:

```text
Hydrogen isotope with mass number 2
```

---

#### Carbon-14

```text
[14C]
```

Meaning:

```text
Radioactive Carbon-14
```

---

### RDKit Example

```python
from rdkit import Chem

mol = Chem.MolFromSmiles("[13C]CO")

for atom in mol.GetAtoms():
    print(atom.GetSymbol(), atom.GetIsotope())
```

Output:

```text
C 13
C 0
O 0
```

---

### Common Function

```python
atom.GetIsotope()
```

Returns:

```text
0  → Normal atom
13 → Carbon-13
2  → Deuterium
14 → Carbon-14
```

---

### Set Isotope

```python
atom.SetIsotope(13)
```

Example:

```python
atom = mol.GetAtomWithIdx(0)
atom.SetIsotope(13)
```

---

### Quick Note

```text
Isotopes are written inside [].

[13C] → Carbon-13
[2H]  → Deuterium
[14C] → Carbon-14

RDKit:
atom.GetIsotope()
atom.SetIsotope()

0 = normal atom
Non-zero = isotope mass number
```

**Rule:** If an isotope is specified in SMILES, it must be written inside square brackets `[]`.
