```python
from rdkit import Chem
from rdkit.Chem import rdMolDescriptors,Draw
```

```python
query1=Chem.MolFromSmarts("[#6&H3]")
mol1=Chem.MolFromSmiles("C")
print(mol1.HasSubstructMatch(query1))
Draw.MolToImage(mol1)
```

```python
mol2=Chem.MolFromSmiles("CO")
print(mol2.HasSubstructMatch(query1))
Draw.MolToImage(mol2)
```

```python
query2=Chem.MolFromSmarts("[#6&H2]")
print(mol1.HasSubstructMatch(query2))
```
