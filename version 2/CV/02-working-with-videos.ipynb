```python
import cv2
```

```python
video=cv2.VideoCapture("Videos/test1.webm")
```

```python
while True:
    ret,frame=video.read()
    if not ret:
        break
    cv2.imshow("Video",frame)
    if cv2.waitKey(1) & 0xFF==ord('q'): # time gap between two frame 1ms
        break
video.release()
cv2.destroyAllWindows()
```

```python
video=cv2.VideoCapture("Videos/test1.webm")
fourcc=cv2.VideoWriter_fourcc(*"mp4v")
height=int(video.get(cv2.CAP_PROP_FRAME_HEIGHT))
width=int(video.get(cv2.CAP_PROP_FRAME_WIDTH))
fps=int(video.get(cv2.CAP_PROP_FPS))
out=cv2.VideoWriter("./Videos/output2.mp4",fourcc,20,(width,height))
```

```python
while True:
    ret,frame=video.read()
    if not ret:
        break
    if cv2.waitKey(1) & 0xFF==ord('q'):
        break
    out.write(frame)
    cv2.imshow("video",frame)
out.release()
video.release()
cv2.destroyAllWindows()
    
```
