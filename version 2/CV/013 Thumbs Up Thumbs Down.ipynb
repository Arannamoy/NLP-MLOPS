```python
import cv2
import cvzone
```

```python
video = cv2.VideoCapture(1)

while True:
    success, frame = video.read()
    if not success:
        break

    results = model(frame, stream=True)

    for r in results:
        for box in r.boxes:
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            w, h = x2 - x1, y2 - y1

            conf = float(box.conf[0])
            cls = int(box.cls[0])
            name = model.names[cls]

            # Green corner box
            cvzone.cornerRect(
                frame,
                (x1, y1, w, h),
                l=30,          # Corner length
                t=5,           # Box thickness
                rt=2,          # Rectangle thickness
                colorR=(0, 255, 0)
            )

            # Fancy label
            cvzone.putTextRect(
                frame,
                f"{name} {conf*100:.1f}%",
                (x1, max(35, y1 - 15)),
                scale=1.6,
                thickness=2,
                colorR=(0, 180, 0),     # Background
                colorT=(255, 255, 255), # Text
                offset=10
            )

    cv2.namedWindow("Video", cv2.WINDOW_NORMAL)
    cv2.imshow("Video", frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

video.release()
cv2.destroyAllWindows()
```
