```python
import cv2
import cvzone
import numpy as np
```

```python
video=cv2.VideoCapture(1)
while True:
    res,frame=video.read()
    if not res:
        break
    frameCanny=cv2.Canny(frame,50,150)
    frameDilated=cv2.dilate(frameCanny,np.ones((5,5),np.uint8),iterations=1)
    imgContours,conFound=cvzone.Utils.findContours(frame,frameDilated,filter=[10,11],drawCon=True)
    frameStack=cvzone.stackImages([frame,imgContours,frameCanny,frameDilated],3,0.5)
    cv2.imshow("Video",frameStack)
    if cv2.waitKey(1) & 0xFF==ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```
