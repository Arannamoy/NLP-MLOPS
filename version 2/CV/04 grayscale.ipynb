```python
import cv2
```

```python
video=cv2.VideoCapture(1)
while True:
    result,frame=video.read()
    if not result:
        break
    grayscale=cv2.cvtColor(frame,cv2.COLOR_BGR2GRAY)
    cv2.imshow("video",grayscale)
    if cv2.waitKey(1) & 0xFF==ord('q'):
        break


video.release()
cv2.destroyAllWindows()
```
