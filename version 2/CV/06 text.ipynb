```python
import cv2
```

```python
video=cv2.VideoCapture(1)

while True:
    res,frame=video.read()
    if not res:
        break
    cv2.putText(frame,"Hello, everyone",(100,100),cv2.FONT_HERSHEY_COMPLEX,2.0,(255,0,0),2)
    cv2.imshow("video",frame)
    if cv2.waitKey(1) & 0xFF==ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```
