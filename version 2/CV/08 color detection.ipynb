```python
import cvzone
import cv2
from cvzone.ColorModule import ColorFinder
```

```python
video=cv2.VideoCapture(1)
colorDetector=ColorFinder(trackBar=True)
hsvValue={}
while True:
    res,frame=video.read()
    if not res:
        break
    frameColor,mask=colorDetector.update(frame,hsvValue)
    cv2.imshow("Color detection",cvzone.stackImages([frame,frameColor,mask],3,0.75))
    if cv2.waitKey(1) & 0xFF==ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```
