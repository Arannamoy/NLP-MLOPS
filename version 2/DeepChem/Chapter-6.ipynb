```python
import deepchem as dc

# Load dataset
tasks, datasets, transformers = dc.molnet.load_tox21()

train_dataset, valid_dataset, test_dataset = datasets

# Create model
model = dc.models.MultitaskClassifier(
    n_tasks=12,
    n_features=1024,
    layer_sizes=[1000, 500],
    dropouts=[0.3, 0.3],
    learning_rate=0.001,
    batch_size=1000
)

# Train
model.fit(train_dataset, nb_epoch=10)

# Metric
metric = dc.metrics.Metric(
    dc.metrics.roc_auc_score,
    np.mean
)

# Evaluate
print("Train:", model.evaluate(train_dataset, [metric], transformers))
print("Valid:", model.evaluate(valid_dataset, [metric], transformers))
print("Test :", model.evaluate(test_dataset, [metric], transformers))
```
