### Ring Closures (`1, 2, 3...`) in SMILES

**Definition:**
Numbers (`1, 2, 3...`) are used to indicate that two atoms are connected to form a **ring**.

---

### Example: Cyclohexane

```text
C1CCCCC1
```

Structure:

```text
   C-C
  /   \
 C     C
 |     |
 C-----C
```

Both `1`s are connected, closing the ring.

---

### Example: Benzene

```text
c1ccccc1
```

Structure:

```text
   ⬡
```

Aromatic 6-membered ring.

---

### Rule

```text
Same number = connect those two atoms
```

Examples:

```text
C1CCCCC1      → Cyclohexane
c1ccccc1      → Benzene
C1CC2CCC1C2   → Two connected rings
```

---

### Quick Note

```text
1,2,3... are ring labels.

First appearance → Ring starts
Second appearance → Ring closes

C1CCCCC1
↑       ↑
same number → bond formed
```

**RDKit:** Ring closure numbers tell the SMILES parser where to create bonds to form cyclic molecular graphs.


````python
from rdkit.Chem import Draw
from rdkit import Chem
````

````python
import cirpy 
smiles=cirpy.resolve("octanol","smiles")
print(smiles)
````

````python
mol=Chem.MolFromSmiles(smiles)
````

````python
for atom in mol.GetAtoms():
    if atom.IsInRing():
        print("Yes")
    else:
        print("No")
````

````python
smiles=cirpy.resolve("benzene","smiles")
mol=Chem.MolFromSmiles(smiles)
for atom in mol.GetAtoms():
    if atom.IsInRing():
        print("Yes")
    else:
        print("No")
````
