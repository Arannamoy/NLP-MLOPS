### RDK Fingerprint (RDKit Fingerprint)

**Definition:**
RDK Fingerprint is a **path-based fingerprint** that encodes molecular paths into a binary vector.

```text id="k7v4m2"
Molecule
   ↓
Atom Paths
   ↓
Hashing
   ↓
Binary Fingerprint
```

---

## Morgan vs RDK

### Morgan Fingerprint

Looks around atoms.

```text id="j3p8n5"
Atom
  ↓
Neighbors
  ↓
Local Environment
```

---

### RDK Fingerprint

Looks at paths.

```text id="t9w2c6"
C - C - O
```

Path:

```text id="x5r1q8"
C→C
C→O
C→C→O
```

These paths are encoded into bits.

---

## RDKit Code

```python id="a4n7k1"
from rdkit import Chem
from rdkit.Chem import RDKFingerprint

mol = Chem.MolFromSmiles("CCO")

fp = RDKFingerprint(mol)

print(fp)
```

---

## Convert to Bit String

```python id="m8q3v2"
print(fp.ToBitString())
```

Output:

```text id="p6k9w4"
001010010101...
```

---

## Default Settings

```python id="r2t7c5"
RDKFingerprint(
    mol,
    minPath=1,
    maxPath=7,
    fpSize=2048
)
```

Meaning:

```text id="u1v8n3"
Path length:
1 → 7 bonds

Fingerprint size:
2048 bits
```

---

## Example

Molecule:

```text id="y4m6k8"
CCO
```

## When to Use?

### Morgan (ECFP)

Best for:

```text id="w8k4m1"
QSAR
DTI
GNN preprocessing
Drug Discovery
```

Most common in papers.

---

### RDK Fingerprint

Best for:

```text id="s3v7n9"
Substructure similarity
Chemical search
Database screening
```
