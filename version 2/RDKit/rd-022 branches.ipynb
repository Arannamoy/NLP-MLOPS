### Branches `()` in SMILES

**Definition:**
Branches represent **side chains** attached to the main chain.

**Syntax:**

```text
()
```

Atoms inside `()` are connected to the atom immediately before the bracket.

---

### Examples

```text
CC(C)C
```

Structure:

```text
    C
    |
C - C - C
```

`(C)` = Carbon branch

---

```text
CC(O)C
```

Structure:

```text
    O
    |
C - C - C
```

`(O)` = Oxygen branch

---

### Multiple Branches

```text
CC(C)(C)O
```

Structure:

```text
    C
    |
C - C - O
    |
    C
```

---

### Rule

```text
Atoms inside () form a branch
from the atom just before the bracket.
```

### Quick Example

```text
CCCC      → Straight chain
CC(C)C    → One branch
CC(C)(C)O → Two branches
```

**RDKit:** `()` creates side-chain connections in the molecular graph.
