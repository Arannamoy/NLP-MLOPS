### Ring Count

Definition:
Number of rings in a molecule.

### RDKit:

rdMolDescriptors.CalcNumRings(mol)

or

mol.GetRingInfo().NumRings()

### Examples:

Benzene      → 1 ring
Naphthalene  → 2 rings

More rings
→ More rigid structure

```python
from rdkit import Chem
from rdkit.Chem import Descriptors,rdMolDescriptors,Draw
```

```python
mol=Chem.MolFromSmiles("C[C@H](N)C(=O)NCC1=CC=C(C=C1)S(=O)(=O)NCCOP(=O)([O-])OCC[NH3+]")
Draw.MolToImage(mol,fitImage=True,size=(900,900))
```

```python
print(rdMolDescriptors.CalcNumRings(mol))
```

```python
print(rdMolDescriptors.CalcNumAliphaticRings(mol))
```
