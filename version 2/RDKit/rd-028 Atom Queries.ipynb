```python
from rdkit import Chem
from rdkit.Chem import rdMolDescriptors,Draw
```

```python
query=Chem.MolFromSmarts("[#6,#8]")
```

```python
Draw.MolToImage(query)
```

```python
rdMolDescriptors.CalcMolFormula(query)
```

```python
mol=Chem.MolFromSmiles("COC")
Draw.MolToImage(mol)
```

```python
mol.HasSubstructMatch(query)
```

```python
mol.GetSubstructMatches(query)
```
