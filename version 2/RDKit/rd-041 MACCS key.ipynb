### MACCS Keys

**Definition:**
MACCS Keys is a predefined molecular fingerprint consisting of **166 structural keys**.

Each bit answers:

```text id="n4q8v1"
Does the molecule contain this feature?

Yes → 1
No  → 0
```

---

## Example

Molecule:

```text id="k7m2p5"
CCO
```

MACCS checks features like:

```text id="r1v9n3"
Has Oxygen?
Has Nitrogen?
Has Aromatic Ring?
Has Halogen?
Has Carbonyl?
...
```

Result:

```text id="y6k4m8"
[0,1,0,1,1,0,...]
```

---

## How It Works?

MACCS uses a fixed dictionary of chemical patterns.

Example:

```text id="t3p7n2"
Key 1  → Feature A
Key 2  → Feature B
...
Key 166 → Feature Z
```

---

## RDKit Code

```python id="c8m5q1"
from rdkit import Chem
from rdkit.Chem import MACCSkeys

mol = Chem.MolFromSmiles("CCO")

fp = MACCSkeys.GenMACCSKeys(mol)
```

---

## Convert to Bits

```python id="v2n8k4"
print(fp.ToBitString())
```

Output:

```text id="j5m1p7"
001010010101...
```

---

## Fingerprint Size

```text id="s9k3n6"
166 bits
```

Always fixed.

Unlike Morgan:

```text id="p4v7m2"
Morgan
→ Custom size (1024, 2048...)

MACCS
→ Fixed size (166)
```

---

## Morgan vs RDK vs MACCS

| Fingerprint | Idea                         |
| ----------- | ---------------------------- |
| Morgan      | Local atom environments      |
| RDK         | Atom paths                   |
| MACCS       | Predefined chemical features |
