### Canonical SMILES

Canonical SMILES is a unique, standardized SMILES representation of a molecule.

A molecule can have multiple valid SMILES strings, but only one canonical SMILES.

Example:

- `CCO`
- `OCC`
- `C(C)O`

All represent ethanol, but their canonical SMILES is:

```text
CCO
```


### One-line memory trick

```text
SMILES = Many ways to write a molecule
Canonical SMILES = One official way to write a molecule
```

### Canonical SMILES does not make different molecules the same. It only gives the same molecule a unique standard representation.

````python
from rdkit import Chem
from torchdrug import data
import matplotlib
%matplotlib inline
````

````python
smiles1=Chem.MolFromSmiles("CCO")
print(Chem.MolToSmiles(smiles1))
mol1=data.Molecule.from_smiles("CCO")
mol1.visualize()
````

````python
smiles2=Chem.MolFromSmiles("OCC")
print(Chem.MolToSmiles(smiles2))
mol2=data.Molecule.from_smiles("OCC")
mol2.visualize()
````
