```python
from rdkit import Chem
from rdkit.Chem import Draw,rdMolDescriptors
```

```python
smiles=Chem.MolFromSmiles("NO")
Draw.MolToImage(smiles)
```

```python
smarts=Chem.MolToSmarts(smiles)
```

```python
o=Chem.MolFromSmiles("O")
```

```python
print(smarts)
```

```python
mol=Chem.MolFromSmarts(smarts)
```

```python
print(mol.HasSubstructMatch(o))
```

```python
print(rdMolDescriptors.CalcMolFormula(o))
```
