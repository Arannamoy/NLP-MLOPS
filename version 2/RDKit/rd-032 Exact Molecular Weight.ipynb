### Descriptors.MolWt() gives average molecular weight,
### while rdMolDescriptors.CalcExactMolWt() gives exact molecular weight based on isotope masses.

```python
from rdkit import Chem
from rdkit.Chem import Descriptors,rdMolDescriptors,Draw
```

```python
mol=Chem.MolFromSmiles("CCCO")
Draw.MolToImage(mol)
```

```python
print(f"Weight of {rdMolDescriptors.CalcMolFormula(mol)}: {rdMolDescriptors.CalcExactMolWt(mol)}")
```
