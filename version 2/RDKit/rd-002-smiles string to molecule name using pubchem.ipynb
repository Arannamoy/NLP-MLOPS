```python
import requests
```

```python
def get_name_from_smiles(smiles):
    url=f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/smiles/{smiles}/property/Title/JSON"
    try:
        response = requests.get(url)
        if response.status_code == 200:
            data = response.json()
            name = data['PropertyTable']['Properties'][0]['Title']
            return name
        else:
            return "Name not found / Custom Molecule"
    except Exception as e:
        return f"Error: {e}"
```

```python
smiles_string = "CC(=O)OC1=CC=CC=C1C(=O)O"
molecule_name = get_name_from_smiles(smiles_string)

print(f"SMILES String: {smiles_string}")
print(f"Actual Name: {molecule_name}")
```
