### Chirality (`@`, `@@`) in SMILES

**Definition:**
Chirality describes the **3D arrangement of atoms** around a stereocenter (usually a carbon atom bonded to 4 different groups).

Two molecules can have the same formula and connectivity but be mirror images of each other.

---

### Chirality Symbols

```text
@
@@
```

These indicate the stereochemistry of an atom.

---

### Example

Lactic Acid:

```text
C[C@H](O)C(=O)O
```

and

```text
C[C@@H](O)C(=O)O
```

These represent two different stereoisomers (mirror images).

---

### Meaning of `@` and `@@`

```text
@   → One stereochemical orientation
@@  → Opposite stereochemical orientation
```

**Important:** `@` does NOT always mean R and `@@` does NOT always mean S.

The actual R/S configuration depends on the atom ordering in the SMILES.

---

### RDKit Example

```python
from rdkit import Chem

mol = Chem.MolFromSmiles("C[C@H](O)C(=O)O")

for atom in mol.GetAtoms():
    print(
        atom.GetIdx(),
        atom.GetSymbol(),
        atom.GetChiralTag()
    )
```

Output:

```text
CHI_UNSPECIFIED
CHI_TETRAHEDRAL_CW
...
```

---

### Check Chirality

```python
atom.GetChiralTag()
```

Possible outputs:

```text
CHI_UNSPECIFIED
CHI_TETRAHEDRAL_CW
CHI_TETRAHEDRAL_CCW
```

---

### Assign R/S Labels

```python
from rdkit.Chem import rdCIPLabeler

mol = Chem.MolFromSmiles("C[C@H](O)C(=O)O")

rdCIPLabeler.AssignCIPLabels(mol)

for atom in mol.GetAtoms():
    if atom.HasProp("_CIPCode"):
        print(atom.GetIdx(), atom.GetProp("_CIPCode"))
```

Output:

```text
1 S
```

or

```text
1 R
```

depending on the molecule.

---

### Quick Visual

```text
      OH
       |
CH3 — C* — H
       |
     COOH
```

`C*` = Chiral Center

The two mirror-image arrangements are different molecules.

````python
from rdkit import Chem

mol = Chem.MolFromSmiles("C[C@H](O)C(=O)O")

# Find chiral centers
print(Chem.FindMolChiralCenters(mol))
````

````python
from rdkit import Chem

mol = Chem.MolFromSmiles("C[C@H](O)C(=O)O")

chiral_centers = Chem.FindMolChiralCenters(mol)

print(Chem.FindMolChiralCenters(mol))

print(type(chiral_centers[0]))

if len(Chem.FindMolChiralCenters(mol)) > 0:
    print("Chiral Molecule")
else:
    print("Achiral Molecule")
````
