```python
from rdkit import Chem
from rdkit.Chem import Draw
import cirpy
```

```python
nitric_acid_smiles=cirpy.resolve("nitric acid","smiles")
benzene_smiles=cirpy.resolve("benzene","smiles")
print(nitric_acid_smiles,benzene_smiles)
```

```python
nitric_acid_mol=Chem.MolFromSmiles(nitric_acid_smiles)
benzene_mol=Chem.MolFromSmiles(benzene_smiles)
```

```python
Draw.MolsToImage([nitric_acid_mol,benzene_mol])
```

```python
Draw.MolToImage(benzene_mol,kekulize=False)
```

```python
for atom in nitric_acid_mol.GetAtoms():
    for neighbor in atom.GetNeighbors():
        print(neighbor.GetAtomicNum())
```
