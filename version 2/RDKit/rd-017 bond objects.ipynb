```python
from rdkit import Chem
from rdkit.Chem import Draw
import cirpy
print(cirpy.resolve("nitric acid","smiles"))
```

```python
mol=Chem.MolFromSmiles("O[N+]([O-])=O")
```

```python
Draw.MolToImage(mol,kekulize=True)
```

```python
for bond in mol.GetBonds():
    print(bond.GetBeginAtom().GetSymbol())
    print(bond.GetEndAtom().GetSymbol())
    print(bond.GetBondDir())
    print(bond.GetIsAromatic())
    print(bond.GetIsConjugated())
    print(bond.GetBondType())
    print(bond.IsInRing())
```
