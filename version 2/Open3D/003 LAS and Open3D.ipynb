#### Coordinate Frame

```python
frame = o3d.geometry.TriangleMesh.create_coordinate_frame(
    size=4.0,
    origin=[0,0,0]
)

o3d.visualization.draw_geometries([pcd, frame])
```

#### Multiple las file

```python
las1=laspy.read("sample_data/sample_0001.las")
las2=laspy.read("sample_data/sample_0002.las")
points1=np.column_stack((las1.x,las1.y,las1.z))
points2=np.column_stack((las2.x,las2.y,las2.z))
```

```python
pcd1=o3d.geometry.PointCloud()
pcd1.points=o3d.utility.Vector3dVector(points1)
pcd2=o3d.geometry.PointCloud()
pcd2.points=o3d.utility.Vector3dVector(points2)

```

```python
frame=o3d.geometry.TriangleMesh.create_coordinate_frame(
    size=4.0,
    origin=[0,0,0]
)
o3d.visualization.draw_geometries([pcd1,pcd2,frame])
```
